```python
# Importação das bibliotecas principais

import pandas as pd  # Manipulação de dados (DataFrame)
import numpy as np   # Operações matemáticas
import matplotlib.pyplot as plt  # Gráficos básicos
import seaborn as sns  # Gráficos estatísticos mais avançados

# Configuração opcional para melhorar visualização
sns.set(style="whitegrid")
```

```python
# Carregar dataset
df = pd.read_csv('ai4i2020.csv')

# Exibir as 5 primeiras linhas do dataset já limpo
df.head()
```

```python
# Informações gerais sobre o dataset

df.info()
```

```python
# Estatísticas básicas do dataset

df.describe()
```

```python
# Verificando se existem valores nulos

df.isnull().sum()
```

```python
# Selecionando apenas colunas numéricas
df_numeric = df.select_dtypes(include=['number'])

# Criando a matriz de correlação
correlation = df_numeric.corr()

# Exibindo
correlation
```

```python
# Criando gráfico de correlação

plt.figure(figsize=(12,6))

sns.heatmap(
    correlation,
    annot=True,     # mostra valores
    cmap='coolwarm' # cores
)

plt.title("Mapa de Correlação entre Variáveis")

plt.show()
```

### Análise de Correlação

Observa-se que variáveis como torque, temperatura e desgaste da ferramenta possuem relação com a falha da máquina.

Essas variáveis são fortes candidatas para o modelo preditivo.

```python
# Detectando outliers com boxplot

plt.figure(figsize=(12,5))

sns.boxplot(data=df)

plt.title("Detecção de Outliers")

plt.show()
```

### Análise de Outliers

Foram identificados valores extremos em algumas variáveis.

Esses valores podem indicar falhas reais ou comportamentos anormais das máquinas.

```python
# Relação entre torque e desgaste

sns.scatterplot(
    x='Torque [Nm]',
    y='Tool wear [min]',
    hue='Machine failure',
    data=df
)

plt.title("Relação entre Torque e Desgaste da Ferramenta")

plt.show()
```

```python
# Distribuição do torque

plt.figure(figsize=(8,4))

df['Torque [Nm]'].hist(bins=30)

plt.title("Distribuição do Torque")

plt.show()
```

#  Validação de Hipóteses

### Hipótese 1:
Quanto maior o torque, maior a chance de falha.  
Confirmada

### Hipótese 2:
O desgaste da ferramenta influencia a falha.  
Confirmada

### Hipótese 3:
Outliers indicam falhas críticas.  
Parcialmente confirmada
