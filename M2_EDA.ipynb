```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
```

```python
import pandas as pd

df = pd.read_csv("ai4i2020.csv")

df.head()
```

## Visão Geral do Dataset

```python
df.info()
df.describe()
```

## Verificação de Valores Ausentes

```python
df.isnull().sum()
```

## Distribuição das Variáveis

```python
df.hist(figsize=(12,8))
plt.show()
```

## Análise de Correlação

```python
df_numeric = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))
sns.heatmap(df_numeric.corr(), annot=True, cmap="coolwarm")
plt.show()
```

## Detecção de Outliers

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

## Relações entre Variáveis

```python
sns.pairplot(df)
plt.show()
```

## Hipóteses Levantadas

- Algumas variáveis apresentam forte correlação entre si.
- Valores extremos podem indicar possíveis falhas operacionais.
- Certos sensores demonstram maior variabilidade, podendo influenciar
o desempenho do sistema.

## Conclusão

A análise exploratória permitiu compreender o comportamento geral
dos sensores, identificar padrões relevantes e possíveis anomalias.
Os resultados obtidos servirão de base para o desenvolvimento
dos modelos de Machine Learning na etapa M3.
